```python
# ==============================================================================
# ACTIVIDAD: INTRODUCCIÓN A PANDAS
# ==============================================================================

# Importar las librerías necesarias
import pandas as pd
import numpy as np

# ------------------------------------------------------------------------------
# 1. Trabajar con datos sintéticos y Crear un DataFrame
# ------------------------------------------------------------------------------
print("--- 1. Creación de DataFrame con Datos Sintéticos ---")

# Configurar una semilla para que los datos aleatorios sean reproducibles
np.random.seed(42)

# Generar datos sintéticos para una tienda de tecnología
n_productos = 5
datos_sinteticos = {
    'Producto': ['Laptop', 'Mouse Wireless', 'Monitor 4K', 'Teclado Mecánico', 'Audífonos Gamer'],
    'Precio_Unitario_USD': [850.00, 25.50, 320.00, 75.00, 60.00],
    'Cantidad_Vendida': np.random.randint(10, 50, size=n_productos), # Cantidades aleatorias entre 10 y 50
    'Stock_Disponible': [15, 100, 25, 40, np.nan] # Introducimos un valor nulo de forma intencional
}

# Crear el DataFrame principal
df_tienda = pd.DataFrame(datos_sinteticos)
print("DataFrame Original:")
display(df_tienda)

# ------------------------------------------------------------------------------
# 2. Hacer operaciones entre columnas
# ------------------------------------------------------------------------------
print("\n--- 2. Operaciones entre Columnas ---")

# Operación 1: Multiplicación de columnas para calcular el Total de Ingresos
df_tienda['Total_Ingresos_USD'] = df_tienda['Precio_Unitario_USD'] * df_tienda['Cantidad_Vendida']

# Operación 2: Calcular el valor estimado del inventario actual
# Nota: Llenamos el valor NaN temporalmente con 0 para evitar errores en el cálculo
df_tienda['Valor_Inventario_Estimado'] = df_tienda['Precio_Unitario_USD'] * df_tienda['Stock_Disponible'].fillna(0)

print("DataFrame con Columnas Calculadas:")
display(df_tienda)

# ------------------------------------------------------------------------------
# 3. Leer un archivo CSV
# ------------------------------------------------------------------------------
print("\n--- 3. Lectura de Archivo CSV ---")

# Creamos un archivo CSV de prueba local en el entorno de Colab para el ejercicio
csv_test_path = 'datos_empleados_prueba.csv'
with open(csv_test_path, 'w') as f:
    f.write("ID,Nombre,Departamento,Sueldo\n1,Ana,Desarrollo,3500\n2,Luis,Diseño,2800\n3,Carlos,Soporte,2200\n")

# Carga y lectura del archivo usando Pandas
try:
    df_csv = pd.read_csv(csv_test_path)
    print(f"Archivo '{csv_test_path}' leído exitosamente:")
    display(df_csv)
except FileNotFoundError:
    print("Error: El archivo no fue encontrado.")
```
